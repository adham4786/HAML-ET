```python
from pathlib import Path

notebook_dir = Path().resolve()
project_root = notebook_dir.parent  # tweak if needed
data_path = project_root / "dataset" / "full.data"

with open(data_path, encoding="utf-8") as f:
    for i, line in enumerate(f, 1):
        parts = line.strip().split(",")
        if len(parts) != 563:
            print("Line", i, "has", len(parts), "columns")
            break
        for j, val in enumerate(parts[:-2], 1):
            try:
                float(val)
            except ValueError:
                print("Line", i, "column", j, "is not numeric:", repr(val))
                break
        else:
            continue
        break
    else:
        print("OK: all lines have 563 columns and numeric features")
```
