### Run dataset through PCA processing

```python
from sklearn.decomposition import PCA
import numpy as np
import pandas as pd
from datetime import datetime as dt


full_df = pd.read_csv(r'../dataset/full.v3_2026-09-24_22:33:20.csv')

df = full_df.drop(columns=['subject', 'Activity'])

pca = PCA(n_components=0.95)

pca_output = pca.fit_transform(df)

df_pca = pd.DataFrame(
    pca_output, columns=[f"PC{i+1}" for i in range(pca_output.shape[1])]
)

df_pca['subject'] = full_df['subject'].values
df_pca['Activity'] = full_df['Activity'].values

now = dt.now().strftime("%Y-%m-%d_%H:%M:%S")

df_pca.to_csv(rf'../dataset/full_v3_pca_{now}.csv', index=False)
```
