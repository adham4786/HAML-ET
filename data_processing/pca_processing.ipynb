### Run dataset through PCA processing

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from datetime import datetime
```

```python
full_df = pd.read_csv("../dataset/full_v4.csv")

now = str(datetime.now().strftime("%Y-%m-%d_%H-%M-%S"))

y = full_df["Activity"]
groups = full_df["subject"]
X = full_df.drop(columns=["subject", "Activity"])

splitter = GroupShuffleSplit(n_splits=1, test_size=0.3, random_state=69)
train_idx, test_idx = next(splitter.split(X, y, groups=groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]

print(X_train.select_dtypes(exclude="number").columns.tolist())
print(X_train.columns.tolist())

pca_pipeline = Pipeline([
    ("pca", PCA(n_components=0.95)),
])

X_train_pca = pca_pipeline.fit_transform(X_train)  # Learn only from training subjects
X_test_pca = pca_pipeline.transform(X_test)         # Apply those same components

columns = [f"PC{i + 1}" for i in range(X_train_pca.shape[1])]

train_pca_df = pd.DataFrame(X_train_pca, columns=columns)
train_pca_df["subject"] = groups.iloc[train_idx].to_numpy()
train_pca_df["Activity"] = y.iloc[train_idx].to_numpy()

test_pca_df = pd.DataFrame(X_test_pca, columns=columns)
test_pca_df["subject"] = groups.iloc[test_idx].to_numpy()
test_pca_df["Activity"] = y.iloc[test_idx].to_numpy()

full_pca_df = pd.concat([train_pca_df, test_pca_df]).sort_index()
full_pca_df.to_csv(f"../dataset/full_v4_pca_{now}.csv", index=False)
```
