## Detect multicollinearity

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../dataset/pca_transformed_data_2026-09-19_15-44-54.csv')

df = df.drop(columns=["subject", "Activity"])

df.head()
```

```python
corr_matrix = df.corr(method="pearson")
corr_matrix.head()
```

```python
# 2. Plot Matrix
plt.figure(figsize=(12, 10))
sns.heatmap(
    corr_matrix, 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    cbar_kws={'label': 'Pearson Correlation'}
)

plt.title('Pearson Correlation Matrix of 71 PCA Components', fontsize=14, fontweight='bold')
plt.xlabel('Principal Components', fontsize=11)
plt.ylabel('Principal Components', fontsize=11)
plt.tight_layout()

plt.show()
```
