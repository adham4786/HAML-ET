## Detect multicollinearity

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../dataset/full_v2.csv')

df = df.drop(columns=["subject", "Activity"])

column_list = df.columns.tolist()

column_list_1 = column_list[:-36]
column_list_2 = column_list[-35:]

df_part_1 = df[column_list_1]
df_part_2 = df[column_list_2]

df_part_1.head()
df_part_2.head()

```

```python
corr_matrix = df_part_1.corr(method="pearson")
```

```python
# Plot corr_matrix
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
corr_matrix = df_part_2.corr(method="pearson")
```

```python
# Plot corr_matrix
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
corr_matrix = df.corr(method="pearson")
```

```python
# Plot corr_matrix
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Correlation Matrix")
plt.show()
```
