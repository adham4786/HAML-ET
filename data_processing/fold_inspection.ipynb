```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupKFold

df = pd.read_csv('../dataset/full_wo_fft.csv')  # Load your dataset

X = df.drop(columns=['Activity', 'subject'])  # Features
y = df['Activity']  # Target variable
groups = df['subject']  # Groups for GroupKFold

gkf = GroupKFold(n_splits=5)  # Initialize GroupKFold with 5 splits

# Loop through folds and inspect unique group IDs in the test set
for fold, (train_index, test_index) in enumerate(gkf.split(X, y, groups)):
    test_groups = np.unique(groups[test_index])
    train_groups = np.unique(groups[train_index])
    
    print(f"Fold {fold + 1}:")
    print(f"  Test group IDs:  {test_groups}")
    print(f"  Train group IDs: {train_groups}\n")
```
