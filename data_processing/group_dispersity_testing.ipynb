```python
import pandas as pd
import numpy as np


df = pd.read_csv('../dataset/activity_distribution.csv')


activity_cols = df.columns.tolist()[1:]  # Exclude the 'Subject' column
k = len(activity_cols)
```

```python
# 1. Sum counts across all subjects
total_counts = df[activity_cols].sum().values

# 2. Compute aggregate probabilities
p_global = total_counts / total_counts.sum()

# 3. Calculate overall normalized Shannon entropy
h_global = -np.sum(p_global * np.log2(p_global))
h_norm_global = h_global / np.log2(k)

print(f"Overall Dataset Normalized Entropy: {h_norm_global:.4f}")
```

```python


# 2. Compute probabilities
counts = df[activity_cols].values
totals = counts.sum(axis=1, keepdims=True)
probs = counts / totals

# 3. Compute Metrics per Subject
# Normalized Shannon Entropy
entropy = -np.sum(probs * np.log2(probs), axis=1)
h_norm = entropy / np.log2(k)

# Herfindahl-Hirschman Index (HHI) & Effective Number of Classes
hhi = np.sum(probs ** 2, axis=1)
n_eff = 1.0 / hhi

# Imbalance Ratio (IR)
ir = counts.max(axis=1) / counts.min(axis=1)

# Compile statistical summary table
metrics_df = pd.DataFrame({
    'Subject': df['Subject'],
    'H_N': h_norm,
    'HHI': hhi,
    'N_eff': n_eff,
    'IR': ir
})

# 4. Print Overall Mathematical Bounds for Paper
print("--- SUMMARY BOUNDS FOR PAPER ---")
print(f"Normalized Entropy (H_N): Mean = {h_norm.mean():.4f}, Min = {h_norm.min():.4f}, Max = {h_norm.max():.4f}")
print(f"HHI Index: Mean = {hhi.mean():.4f}, Min = {hhi.min():.4f}, Max = {hhi.max():.4f}")
print(f"Effective Classes (N_eff): Mean = {n_eff.mean():.2f}, Min = {n_eff.min():.2f}, Max = {n_eff.max():.2f}")
print(f"Imbalance Ratio (IR): Mean = {ir.mean():.2f}, Min = {ir.min():.2f}, Max = {ir.max():.2f}")

# 5. Export LaTeX Table for Paper Draft
latex_output = metrics_df.describe().T[['mean', 'std', 'min', '50%', 'max']].to_latex(
    float_format="%.3f",
    caption="Statistical Dispersion Metrics Disproving Single-Class Dominance Across 30 Subjects",
    label="tab:dispersion_metrics"
)
print("\n--- LATEX TABLE FOR PAPER ---")
print(latex_output)
```
